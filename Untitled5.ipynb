## 1. Import Library yang Diperlukan

**Penjelasan Sederhana:**
Sebelum kita bekerja, kita perlu menyiapkan "alat-alat kerja" kita (disebut sebagai *library*).
*   `pandas` (alias `pd`): Seperti Excel, digunakan untuk mengolah data dalam bentuk tabel.
*   `numpy` (alias `np`): Digunakan untuk operasi matematika dan angka.
*   `matplotlib` & `seaborn` (alias `plt` & `sns`): Alat untuk menggambar grafik/visualisasi data agar cantik dan mudah dibaca.
*   `datetime`: Untuk mengolah data tanggal dan waktu.
*   `sklearn` (Scikit-Learn): Library khusus Machine Learning. Di sini kita mengambil alat standardisasi (`StandardScaler`) dan algoritma pengelompokan (`KMeans`).

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import datetime as dt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Mengatur visualisasi agar terlihat rapi
%matplotlib inline
sns.set(style="whitegrid")
import warnings
warnings.filterwarnings('ignore')
````

## 2. Mengunduh dan Membaca Dataset

**Penjelasan Sederhana:**
Di tahap ini, kode akan otomatis mengunduh file data penjualan (`OnlineRetail.csv`) dari internet (Kaggle) menggunakan bantuan `kagglehub`.
Setelah diunduh, kita membacanya menggunakan Pandas (`pd.read_csv`) dan menyimpannya dalam variabel bernama `df` (singkatan dari *DataFrame*, atau tabel data kita).

````python
import kagglehub
import os

# Mengunduh versi terbaru dari dataset Kaggle
path = kagglehub.dataset_download("hellbuoy/online-retail-customer-clustering")
print("Path ke file dataset:", path)

# Menemukan file OnlineRetail.csv di folder hasil unduhan
csv_path = os.path.join(path, 'OnlineRetail.csv')

# Membaca dataset dengan encoding ISO-8859-1
df = pd.read_csv(csv_path, encoding='ISO-8859-1')
````

## 3. Eksplorasi Data Awal (Mengenal Data Anda)

**Penjelasan Sederhana:**
Sebelum mengolah data, kita harus berkenalan dulu dengan data kita:
*   `df.head()`: Menampilkan 5 baris pertama data agar kita tahu wujud tabelnya.
*   `df.shape`: Memberitahu kita berapa total baris dan kolom yang ada.
*   `df.info()`: Memperlihatkan jenis data di setiap kolom (apakah angka, teks, atau tanggal) dan apakah ada data yang kosong (Null).
*   `df.describe()`: Memberikan ringkasan statistik seperti rata-rata belanja, jumlah barang paling sedikit/banyak, dll.

````python
print("=== 5 Data Pertama ===")
display(df.head())

print("\n=== Ukuran Dataset ===")
print(f"Baris: {df.shape[0]}, Kolom: {df.shape[1]}")

print("\n=== Informasi Kolom ===")
df.info()

print("\n=== Statistik Deskriptif ===")
display(df.describe())
````

## 4. Pembersihan Data (Data Cleaning)

**Penjelasan Sederhana:**
Data dari dunia nyata seringkali kotor. Di sini kita membersihkannya agar hasil analisis akurat:
1.  **Hapus CustomerID Kosong**: Kita tidak bisa mengelompokkan pelanggan jika kita tidak tahu ID mereka.
2.  **Hapus Duplikat**: Menghapus baris transaksi yang tercatat ganda secara tidak sengaja.
3.  **Hapus Pembatalan**: Transaksi batal ditandai dengan huruf 'C' (*Cancelled*) pada nomor faktur, kita buang ini karena tidak mencerminkan pembelian asli.
4.  **Hapus Angka Minus**: Jumlah barang (`Quantity`) dan harga (`UnitPrice`) harus lebih besar dari 0 agar masuk akal.

````python
# 1. Menghapus data dengan CustomerID kosong
df = df.dropna(subset=['CustomerID'])
df['CustomerID'] = df['CustomerID'].astype(str)

# 2. Menghapus data duplikat
df = df.drop_duplicates()

# 3. Menghapus transaksi pembatalan (InvoiceNo diawali 'C')
df = df[~df['InvoiceNo'].str.contains('C', na=False)]

# 4. Memastikan Quantity dan UnitPrice bernilai positif (> 0)
df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0)]

print(f"Ukuran dataset setelah dibersihkan: {df.shape[0]} baris, {df.shape[1]} kolom.")
````

## 5. Visualisasi Sederhana (Melihat Transaksi Tiap Negara)

**Penjelasan Sederhana:**
Kita membuat grafik batang (*bar plot*) untuk melihat dari negara mana saja pelanggan kita berasal. Ini membantu kita melihat pasar terbesar toko online kita secara instan.

````python
plt.figure(figsize=(12, 6))
top_countries = df['Country'].value_counts().head(10)
sns.barplot(x=top_countries.values, y=top_countries.index, palette='viridis')
plt.title('10 Negara dengan Jumlah Transaksi Terbanyak')
plt.xlabel('Jumlah Transaksi')
plt.ylabel('Negara')
plt.show()
````

## 6. Pembuatan Fitur RFM (Recency, Frequency, Monetary)

**Penjelasan Sederhana:**
Di sinilah inti dari segmentasi pelanggan. Kita mengubah data transaksi harian yang ribuan jumlahnya menjadi hanya 3 angka sakti untuk setiap pelanggan:
1.  **Recency (Kekinian)**: Dihitung dengan rumus `tanggal analisis - tanggal belanja terakhir`. Menghasilkan angka: "Sudah berapa hari pelanggan ini tidak belanja?"
2.  **Frequency (Kerapian)**: Menghitung berapa kali ID pelanggan tersebut muncul dengan nomor faktur yang unik. Menghasilkan angka: "Berapa kali mereka belanja?"
3.  **Monetary (Uang)**: Mengalikan jumlah barang dengan harga, lalu dijumlahkan semuanya per pelanggan. Menghasilkan angka: "Berapa banyak uang yang sudah mereka habiskan di toko kita?"

````python
# Menghitung Total Belanja (Monetary) per baris
df['TotalAmount'] = df['Quantity'] * df['UnitPrice']

# Mengonversi format kolom InvoiceDate menjadi datetime
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], format='%d-%m-%Y %H:%M', errors='coerce')
# Jika format di atas menghasilkan banyak NaT, coba auto-parse:
if df['InvoiceDate'].isnull().sum() > len(df) * 0.5:
    df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

# Menentukan tanggal acuan analisis (1 hari setelah tanggal transaksi terakhir dalam dataset)
max_date = df['InvoiceDate'].max()
analysis_date = max_date + dt.timedelta(days=1)

# Membuat dataframe RFM
rfm = df.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (analysis_date - x.max()).days,  # Recency
    'InvoiceNo': 'nunique',                                   # Frequency
    'TotalAmount': 'sum'                                      # Monetary
}).reset_index()

# Mengubah nama kolom agar representatif
rfm.columns = ['CustomerID', 'Recency', 'Frequency', 'Monetary']
display(rfm.head())
````

## 7. Penanganan Outlier dan Standardisasi Data

**Penjelasan Sederhana:**
*   **Outlier (Pencilan)**: Adalah pelanggan yang sifat belanjanya terlalu ekstrem (misalnya, sultan yang belanja miliaran rupiah sekali bayar). Pelanggan ekstrem ini dibuang sementara agar tidak merusak perhitungan rata-rata kelompok.
*   **Standardisasi (StandardScaler)**: Karena satuan RFM berbeda (Recency = hari, Frequency = kali, Monetary = rupiah), kita harus menyamakan skalanya menjadi rentang yang setara (biasanya berkisar antara -3 sampai +3). Jika tidak disamakan, nilai rupiah yang besar akan mendominasi algoritma clustering.

````python
# Penanganan Outlier menggunakan IQR
columns = ['Recency', 'Frequency', 'Monetary']
for col in columns:
    Q1 = rfm[col].quantile(0.05)
    Q3 = rfm[col].quantile(0.95)
    IQR = Q3 - Q1
    rfm = rfm[(rfm[col] >= Q1 - 1.5*IQR) & (rfm[col] <= Q3 + 1.5*IQR)]

# Melakukan Standardisasi (Normalisasi)
scaler = StandardScaler()
rfm_scaled = scaler.fit_transform(rfm[['Recency', 'Frequency', 'Monetary']])

# Membuat DataFrame hasil standardisasi
rfm_scaled_df = pd.DataFrame(rfm_scaled, columns=['Recency', 'Frequency', 'Monetary'])
display(rfm_scaled_df.head())
````

## 8. Menentukan Jumlah Cluster Optimal dengan Elbow Method

**Penjelasan Sederhana:**
Bagaimana kita tahu harus membagi pelanggan menjadi 3, 4, atau 5 kelompok?
Kita gunakan **Metode Siku (Elbow Method)**. Kode ini mencoba membuat kelompok dari jumlah 1 sampai 10, lalu mengukur seberapa jauh jarak pelanggan ke pusat kelompoknya (disebut *Inersia* atau *WCSS*).

Ketika digambarkan ke grafik, kita cari titik di mana grafik tersebut mulai landai (membentuk lekukan mirip siku tangan). Titik siku itulah jumlah kelompok (K) terbaik.

````python
wcss = []
K_range = range(1, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, init='k-means++', random_state=42)
    kmeans.fit(rfm_scaled_df)
    wcss.append(kmeans.inertia_)

# Visualisasi Elbow Method
plt.figure(figsize=(10, 5))
plt.plot(K_range, wcss, marker='o', linestyle='--', color='b')
plt.title('Elbow Method untuk Menentukan Jumlah Cluster Optimal')
plt.xlabel('Jumlah Cluster (K)')
plt.ylabel('WCSS (Inersia)')
plt.xticks(K_range)
plt.show()
````

## 9. Menerapkan K-Means Clustering

**Penjelasan Sederhana:**
Setelah tahu jumlah kelompok terbaik (di sini kita pilih 3), kita perintahkan komputer untuk mengelompokkan pelanggan kita ke salah satu dari 3 kelompok tersebut.
Hasil pengelompokan ini disimpan sebagai label angka (0, 1, atau 2) di kolom baru bernama `Cluster`.

````python
# Tentukan K berdasarkan visualisasi Elbow di atas (Misal: K=3)
optimal_k = 3

kmeans = KMeans(n_clusters=optimal_k, init='k-means++', random_state=42)
rfm['Cluster'] = kmeans.fit_predict(rfm_scaled_df)

print("Proses clustering selesai. Label cluster telah ditambahkan ke dataframe.")
````

## 10. Mengukur Kualitas Pengelompokan (Silhouette Score)

**Penjelasan Sederhana:**
*Silhouette Score* adalah nilai matematika (antara -1 hingga 1) untuk menilai seberapa bagus hasil kerja K-Means:
*   Mendekati **1**: Kelompok terbagi sangat rapi dan tidak saling tumpang tindih.
*   Mendekati **0**: Kelompoknya agak membingungkan atau saling menempel.

Skor sekitar 0.50 ke atas menunjukkan hasil pengelompokan yang sudah cukup baik dan logis.

````python
sil_score = silhouette_score(rfm_scaled_df, rfm['Cluster'])
print(f"Silhouette Score untuk {optimal_k} Cluster adalah: {sil_score:.4f}")
````

## 11. Visualisasi Hasil Pengelompokan (Scatter Plot)

**Penjelasan Sederhana:**
Kita menggambar titik-titik pelanggan pada grafik 2 dimensi. Warna titik mewakili nomor kelompoknya (Cluster).
Dari grafik ini, kita bisa melihat secara visual bahwa pelanggan dengan warna yang sama berkumpul di area yang sama, membuktikan bahwa K-Means telah bekerja dengan baik.

````python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Scatter Plot: Recency vs Frequency
sns.scatterplot(data=rfm, x='Recency', y='Frequency', hue='Cluster', palette='Set1', ax=axes[0])
axes[0].set_title('Recency vs Frequency')

# Scatter Plot: Frequency vs Monetary
sns.scatterplot(data=rfm, x='Frequency', y='Monetary', hue='Cluster', palette='Set1', ax=axes[1])
axes[1].set_title('Frequency vs Monetary')

plt.suptitle('Visualisasi Hasil K-Means Clustering', fontsize=16)
plt.show()
````

## 12. Menghitung Jumlah Pelanggan per Kelompok

**Penjelasan Sederhana:**
Kode ini menghitung berapa banyak total pelanggan yang masuk ke kelompok 0, kelompok 1, dan kelompok 2. Ini penting agar kita tahu kelompok mana yang anggotanya paling mayoritas.

````python
cluster_counts = rfm['Cluster'].value_counts().sort_index()
print("Jumlah pelanggan pada setiap cluster:")
for cluster, count in cluster_counts.items():
    print(f"Cluster {cluster}: {count} pelanggan")
````

## 13. Analisis Nilai Rata-rata RFM Tiap Kelompok

**Penjelasan Sederhana:**
Untuk mengetahui perilaku setiap kelompok, kita hitung nilai rata-rata *Recency*, *Frequency*, dan *Monetary* mereka. Tabel ini akan menjadi dasar kita untuk membaca karakter atau kepribadian belanja dari masing-masing kelompok pelanggan.

````python
analysis_rfm = rfm.groupby('Cluster').agg({
    'Recency': 'mean',
    'Frequency': 'mean',
    'Monetary': 'mean'
}).reset_index()

display(analysis_rfm)
````

## 14. Interpretasi Karakteristik Setiap Kelompok

**Penjelasan Sederhana:**
Kode ini menerjemahkan angka-angka rata-rata pada tahap sebelumnya menjadi kalimat bisnis yang mudah dipahami manusia:
*   **Pelanggan Loyal (Sultan)**: Belanja baru-baru ini, sangat sering, dan habis uang banyak.
*   **Pelanggan Potensial (Biasa)**: Belanja sedang-sedang saja, masih aktif.
*   **Pelanggan Pasif (Hampir Hilang)**: Sudah lama sekali tidak terlihat belanja, frekuensi sedikit, dan belanja sedikit.

````python
# Mengurutkan cluster berdasarkan performa Monetary untuk pemetaan label yang logis
sorted_clusters = analysis_rfm.sort_values(by='Monetary', ascending=False).reset_index(drop=True)

print("=== Interpretasi Karakteristik Pelanggan ===")
for idx, row in sorted_clusters.iterrows():
    c_id = int(row['Cluster'])
    rec = row['Recency']
    freq = row['Frequency']
    mon = row['Monetary']

    if idx == 0:
        print(f"\nCluster {c_id}: Pelanggan Loyal / Bernilai Tinggi (Loyal & High Value Customers)")
        print(f"  - Karakteristik: Baru saja bertransaksi (Recency rendah: {rec:.1f} hari), sangat sering belanja (Frequency tinggi: {freq:.1f} kali), dan total belanja besar (Monetary tinggi: {mon:.1f}).")
    elif idx == 1:
        print(f"\nCluster {c_id}: Pelanggan Potensial (Potential / Warm Customers)")
        print(f"  - Karakteristik: Tingkat keaktifan sedang (Recency: {rec:.1f} hari), frekuensi belanja menengah ({freq:.1f} kali), dan pengeluaran cukup ({mon:.1f}).")
    else:
        print(f"\nCluster {c_id}: Pelanggan Pasif / Berisiko Pergi (At-Risk / Churned Customers)")
        print(f"  - Karakteristik: Sudah lama tidak belanja (Recency tinggi: {rec:.1f} hari), frekuensi belanja sangat rendah ({freq:.1f} kali), dan total belanja paling sedikit ({mon:.1f}).")
````
